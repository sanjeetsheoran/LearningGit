### K-Nearest Neighbors (KNN) Classification

Let's apply K-Nearest Neighbors (KNN) classification to a simple dataset where we predict 'pass' or 'fail' based on 'study hours'.

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Generate Sample Data for Study Hours and Pass/Fail
# Let's create a hypothetical dataset
np.random.seed(42)
study_hours = np.random.rand(50) * 10 # 0 to 10 hours of study

# Define a simple rule for pass/fail based on study hours
# For example, more than 5 hours is likely a pass
pass_fail = (study_hours > 5).astype(int) # 1 for pass, 0 for fail

data = pd.DataFrame({'Study_Hours': study_hours, 'Pass_Fail': pass_fail})
display(data.head())

# Prepare data for sklearn
X = data[['Study_Hours']]
y = data['Pass_Fail']
```

```python
# 2. Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Initialize and train the KNN Classifier
# We'll start with n_neighbors=3
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

print("KNN Model trained successfully!")
```

```python
# 4. Make predictions
y_pred = knn.predict(X_test)

# 5. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:")
print(report)

# Example prediction for a new student
new_student_hours = np.array([[7.5]]) # A student studied 7.5 hours
prediction = knn.predict(new_student_hours)

if prediction[0] == 1:
    print(f"\nPrediction for a student studying {new_student_hours[0][0]} hours: Pass")
else:
    print(f"\nPrediction for a student studying {new_student_hours[0][0]} hours: Fail")
```
